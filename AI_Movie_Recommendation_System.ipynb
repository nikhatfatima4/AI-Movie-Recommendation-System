```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("Libraries loaded successfully!")
```

```python
url = "https://raw.githubusercontent.com/vamshi121/TMDB-5000-Movie-Dataset/main/tmdb_5000_movies.csv"

movies = pd.read_csv(url)

print("Dataset loaded successfully!")
print(movies.shape)
```

```python
movies.head()
```

```python
movies = movies[['id', 'title', 'genres', 'keywords', 'overview']]

movies.head()
```

```python
movies['tags'] = (
    movies['genres'].fillna('') + ' ' +
    movies['keywords'].fillna('') + ' ' +
    movies['overview'].fillna('')
)

movies[['title', 'tags']].head()
```

```python
vectorizer = CountVectorizer(stop_words='english')

vectors = vectorizer.fit_transform(movies['tags'])

print("Text converted successfully!")
print(vectors.shape)
```

```python
similarity = cosine_similarity(vectors)

print("Similarity calculated successfully!")
```

```python
def recommend(movie):
    movie = movie.lower()

    matches = movies[movies['title'].str.lower() == movie]

    if matches.empty:
        print("Movie not found.")
        return

    index = matches.index[0]

    distances = sorted(
        list(enumerate(similarity[index])),
        reverse=True,
        key=lambda x: x[1]
    )

    print(f"\nBecause you liked '{movies.iloc[index]['title']}', you may also like:\n")

    for i in distances[1:6]:
        print("•", movies.iloc[i[0]]['title'])
```

```python
recommend("Avatar")
```

```python
recommend("The Dark Knight")
```

```python
recommend("Titanic")
```

```python
recommend("Inception")
```

```python
recommend("Jurassic Park")
```
